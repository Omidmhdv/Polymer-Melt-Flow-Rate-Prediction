In particular, as described on the web page above:

Polymer properties such as density, melt index, and melt flow rate must be kept within tight specifications for each grade. This case study is to analyze polymer production data to predict melt flow rate.

**Background:** There are gas phase and liquid slurry reactors that create polymers (polyethylene, polypropylene, polystyrene, and others) from chemical building blocks known as monomers (C2=, C3=, C4=, iC5=, and others). A catalyst is injected with the monomers under carefully controlled temperature and pressure conditions to cause a reaction that grows the polymer chains. Hydrogen is a chain transfer agent to stop the growth of the polymer chain. If the polymer chains grow too long then the polymer is too viscous for manufacturing in films, injection molding, or other applications.

This case study focuses on measurements of Melt Flow Rate (MFR) to determine the polymer viscosity based on reactor conditions. An accurate model is desirable so that the infrequent lab samples (every 2-8 hours) are supplemented with a virtual and continuous "soft sensor". A model that runs in real-time simulation alongside the physical reactor is called a digital twin.

## **Question 6** (10 points)

Pick one or two different functions and apply the to `MFR` and see if a transformed `MFR` is easier to predict.  Does transforming `MFR` change what inputs give good predictions?

__Answer__:  We applied the square root (sqrtMFR) and inverse (invMFR) transformations to the MFR target, in addition to the natural logarithm (lnMFR). After evaluating linear regression models using all available features ('C3', 'H2R', 'Pressure', 'Level', 'C2', 'Cat', 'Temp'), we observed the following:

Ease of Prediction (R^2): When comparing the R^2 values, the original MFR (R^2: 0.5233) yielded the highest coefficient of determination, meaning it explained the most variance in the data with this linear model and feature set. sqrtMFR (R^2: 0.5208) was very close, followed by lnMFR (R^2: 0.5113), and then invMFR (R^2: 0.4807). This suggests that, for this specific set of features and a linear model, none of the tested transformations made MFR inherently 'easier to predict' in terms of explaining a greater proportion of its variance compared to the untransformed MFR.
However, the MSE values become significantly lower.This indicates that a model’s predictions are closer to the actual, observed values, signifying higher accuracy.

````python
import numpy as np
from sklearn import linear_model
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# Apply additional transformations to MFR in the 'data' DataFrame
data['sqrtMFR'] = np.sqrt(data['MFR'])
data['invMFR'] = 1 / data['MFR']

# Re-split the 'data' DataFrame (which now includes transformed MFR columns)
train_data, test_data = train_test_split(data, test_size=0.2, shuffle=True, random_state=1)

# Remove rows with any NaN values from the train and test splits
train_dropna = train_data.dropna()
test_dropna = test_data.dropna()

# Define features to be used for the models
MFRM=['MFR','lnMFR','invMFR','sqrtMFR']

for MFRM in MFRM:
  method = linear_model.LinearRegression()
  model = method.fit(
  X=train_dropna[['C3', 'H2R', 'Pressure', 'Level', 'C2', 'Cat', 'Temp']].to_numpy(),
  y=train_dropna[MFRM].to_numpy())

  r2 = method.score(test_dropna[['C3', 'H2R', 'Pressure', 'Level', 'C2', 'Cat', 'Temp']].to_numpy(),test_dropna[MFRM].to_numpy())
  mse = mean_squared_error(method.predict(test_dropna[['C3', 'H2R', 'Pressure', 'Level', 'C2', 'Cat', 'Temp']].to_numpy()), test_dropna[MFRM].to_numpy())

  print(MFRM + '  '+'R^2: ' + str(r2))
  print(MFRM +'  '+'mse: ' + str(mse))
  print('------------------------')
````

## **Question 7** (10 points)

Just like using transformations on the target (`MFR` in this case), you can use transformations on the inputs. Can you predict a transformed `MFR` from a set of transformed inputs?

Ok, things start to explode now with so many combinations. Just try a few :-)

__Answer__: With tranformation of the input data, we don't osserve significant change in R2 and mse. However, tranformation of the MFR can give different R2 and mse.


````python
import numpy as np
from sklearn import linear_model
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

# Apply additional transformations to MFR in the 'data' DataFrame
data['sqrtC3'] = np.sqrt(data['C3'])
data['invC3'] = 1 / data['C3']
C3M=['C3','sqrtC3','invC3']

# Apply additional transformations to MFR in the 'data' DataFrame
data['sqrtPressure'] = np.sqrt(data['Pressure'])
data['invPressure'] = 1 / data['Pressure']
PressureM=['Pressure','sqrtPressure','invPressure']

# Re-split the 'data' DataFrame (which now includes transformed MFR columns)
train_data, test_data = train_test_split(data, test_size=0.2, shuffle=True, random_state=1)

# Remove rows with any NaN values from the train and test splits
train_dropna = train_data.dropna()
test_dropna = test_data.dropna()

# Define features to be used for the models
MFRM=['MFR']

for MFRM in MFRM:
  for C3M in C3M:
    method = linear_model.LinearRegression()
    model = method.fit(
    X=train_dropna[[C3M, 'H2R', 'Pressure', 'Level', 'C2', 'Cat', 'Temp']].to_numpy(),
    y=train_dropna[MFRM].to_numpy())

    r2 = method.score(test_dropna[[C3M, 'H2R', 'Pressure', 'Level', 'C2', 'Cat', 'Temp']].to_numpy(),test_dropna[MFRM].to_numpy())
    mse = mean_squared_error(method.predict(test_dropna[[C3M, 'H2R', 'Pressure', 'Level', 'C2', 'Cat', 'Temp']].to_numpy()), test_dropna[MFRM].to_numpy())

    print(MFRM +' '+ C3M +' '+'R^2: ' + str(r2))
    print(MFRM +' '+ C3M +' '+'mse: ' + str(mse))
    print('------------------------')

for PressureM in PressureM:
  method = linear_model.LinearRegression()
  model = method.fit(
  X=train_dropna[['C3', 'H2R', PressureM, 'Level', 'C2', 'Cat', 'Temp']].to_numpy(),
  y=train_dropna['MFR'].to_numpy())

  r2 = method.score(test_dropna[['C3', 'H2R', PressureM, 'Level', 'C2', 'Cat', 'Temp']].to_numpy(),test_dropna['MFR'].to_numpy())
  mse = mean_squared_error(method.predict(test_dropna[['C3', 'H2R', PressureM, 'Level', 'C2', 'Cat', 'Temp']].to_numpy()), test_dropna['MFR'].to_numpy())

  print(MFRM +' '+ PressureM +' '+'R^2: ' + str(r2))
  print(MFRM +' '+ PressureM +' '+'mse: ' + str(mse))
  print('------------------------')
````
